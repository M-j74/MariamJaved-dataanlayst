```python


from scipy import stats

from sklearn.model_selection import train_test_split, cross_val_score, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.pipeline import Pipeline
```

```python
df['category'].value_counts()
```

```python
bikes_margin = df[df['category'] == 'Bikes']['profit_margin']
components_margin = df[df['category'] == 'Components']['profit_margin']

t_stat, p_value = stats.ttest_ind(bikes_margin, components_margin)
print('t_test:',t_stat,'p_values:',p_value)
```

```python
#we can reject the null hypothisis becuse there is a margin in category
```

```python
df['territory'].value_counts()
```

```python
groups = [group['LineTotal'].values for _, group in df.groupby('territory')]
f_stat, p_value = stats.f_oneway(*groups)
print('F-statistic:', f_stat, 'p-value:', p_value)
```

```python
#here we get also reject the null hypothesis becuse territory contribute diffrently
```

```python
contingency_table = pd.crosstab(df['category'], df['territory'])
chi2_stat, p_value, dof, expected = stats.chi2_contingency(contingency_table)
print(chi2_stat, p_value, dof)
```

```python
#here we reject null hypothis
```

```python
x = pd.get_dummies(df[['category', 'territory']])
x = pd.concat([x, df[['OrderQty', 'UnitPrice', 'ListPrice', 'StandardCost']]], axis=1)
y = df['LineTotal']

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
model= LinearRegression()
model.fit(x_train,y_train)
yhat=model.predict(x_test)
```

```python
MSE=mean_squared_error
print('mean_error:',MSE(y_test,yhat))
```

```python
r2=r2_score
r2_linear = r2(y_test, yhat)
print('pipeline r2 :', r2_linear)
```

```python
#we get 0.6 r_score mena it not much but enough solid fo linear model
```

```python
input = Pipeline([('scale',StandardScaler()),('model',LinearRegression())])
input.fit(x_train,y_train)
yhat_pipeline=input.predict(x_test)
```

```python
mse_scaler=print('pipeline MSE:',MSE(y_test,yhat_pipeline))
```

```python
r2_scaler=print('pipeline r2 :',r2(y_test,yhat_pipeline))
```

```python
r2_scaler = r2(y_test, yhat_pipeline)
print('pipeline r2 :', r2_scaler)
```

```python
r2_linear = float(r2_linear)
r2_scaler = float(r2_scaler)
r2_scores = pd.Series([r2_linear, r2_scaler], index=['linear', 'scaler'])
r2_scores.plot(kind='bar', title='Comparison of r2_linear vs r2_scaler')
plt.show()
```

```python
linear=LinearRegression()
RFR=RandomForestRegressor()
```

```python
cvs_l=cross_val_score(linear,x_train,y_train,cv=5)
```

```python
cvs_l_mean=cvs_l.mean()
```

```python
cvs_l_std=cvs_l.std()
```

```python
cvs_means=[cvs_l_mean,cvs_rfr_mean]
cvs_stds=[cvs_l_std,cvs_rfr_std]
```

```python
import matplotlib.pyplot as plt
import numpy as np

models = ['Linear Regression', 'Random Forest']
x_pos = np.arange(len(models))

fig, axes = plt.subplots(1, 2, figsize=(10, 4))

# Plot 1: Mean scores
axes[0].bar(x_pos, cvs_means, color=['skyblue', 'lightgreen'])
axes[0].set_xticks(x_pos)
axes[0].set_xticklabels(models)
axes[0].set_ylabel('Mean CV Score')
axes[0].set_title('Cross‑Validation Mean Scores')

# Plot 2: Standard deviations
axes[1].bar(x_pos, cvs_stds, color=['lightcoral', 'gold'])
axes[1].set_xticks(x_pos)
axes[1].set_xticklabels(models)
axes[1].set_ylabel('CV Score Std')
axes[1].set_title('Cross‑Validation Score Variability')

plt.tight_layout()
plt.show()

```

```python

rfr = RandomForestRegressor(random_state=42)


```

```python
param_grid = {
    'n_estimators': [50, 100, 200],
    'max_depth': [None, 5, 10, 20]
}

```

```python

grid_search = GridSearchCV(
    estimator=rfr,
    param_grid=param_grid,
    cv=3,
    scoring='r2',
    n_jobs=-1
)


grid_search.fit(x_train, y_train)


print("Best parameters:", grid_search.best_params_)
print("Best R² score:", grid_search.best_score_)
```

```python


# Get feature importances from the best Random Forest model
importances = grid_search.best_estimator_.feature_importances_

# Create a DataFrame for easy plotting
feature_importance_df = pd.DataFrame({
    'Feature': x_train.columns,
    'Importance': importances
}).sort_values(by='Importance', ascending=False)

# Plot
plt.figure(figsize=(8, 5))
plt.barh(feature_importance_df['Feature'], feature_importance_df['Importance'], color='teal')
plt.xlabel('Importance Score')
plt.ylabel('Features')
plt.title('Random Forest Feature Importance')
plt.gca().invert_yaxis()  # Highest importance at top
plt.show()
```
