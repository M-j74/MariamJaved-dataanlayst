```python


col_names = ['CustomerId','FirstName','LastName','Country','InvoiceId',
             'BillingCountry','Total','InvoiceDate','Quantity','UnitPrice',
             'track_name','Milliseconds','album_title','artist_name','genre_name']

df = pd.read_csv(r"C:\Users\micro\OneDrive\Documents\chinookmaster_csv.csv", header=None, names=col_names)
df.head()
```

```python
df.head()
```

```python
df.columns
```

```python
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
```

```python
df.info()
```

```python
df['Invoiceyear'] =df['InvoiceDate'].dt.year
df['InvoiceMonth'] =df['InvoiceDate'].dt.month
```

```python
df.info()
```

```python
rev_by_count_ = df.groupby('BillingCountry')['Total'].sum().nlargest(10)
```

```python
import seaborn as ans
import matplotlib.pyplot as plt
```

```python
rev_by_count_.plot(kind='bar', figsize=(10,5), title='Top 10 Countries by Revenue')
plt.xlabel('Country')
plt.ylabel('Total Revenue')
plt.show()
```

```python
# Step 1: Calculate revenue
df['Revenue'] = df['Quantity'] * df['UnitPrice']

# Step 2: Group by Genre (make sure Genre column exists!)
genre_revenue = df.groupby('genre_name')['Revenue'].sum().nlargest(10)

# Step 3: Plot with labels
genre_revenue.plot(kind='bar', figsize=(10,5), title='Top 10 Genres by Revenue')
plt.xlabel('Genre')
plt.ylabel('Total Revenue')
plt.show()
```

```python
monthly_sales = df.groupby('InvoiceMonth')['Revenue'].sum()

# Step 4: Plot line chart
monthly_sales.plot(kind='line', figsize=(10,5), title='Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Revenue')
plt.show()
```

```python
monthly_sales = df.groupby('Invoiceyear')['Revenue'].sum()

# Step 4: Plot line chart
monthly_sales.plot(kind='line', figsize=(10,5), title='Monthly Sales Trend')
plt.xlabel('year')
plt.ylabel('Total Revenue')
plt.show()
```

```python
artist_revenue = df.groupby('artist_name')['Total'].sum().nlargest(10)
artist_revenue.plot(kind='barh', figsize=(10,5), title='revenue by artist')
plt.xlabel('artist_name')
plt.ylabel('Total Revenue')
plt.show()
```

```python
milli_to_min = df['Milliseconds']/(1000*60)
```

```python
milli_to_min = df['Milliseconds'] / (1000 * 60)
milli_to_min.plot(
    kind='hist',
    bins=20,
    color='skyblue',
    edgecolor='black',
    figsize=(10,5),
    title='Track Length Distribution'
)
plt.xlabel('Track Length (minutes)')
plt.ylabel('Frequency')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.scatterplot(x='Quantity', y='UnitPrice', data=df, color='red', marker='*')
plt.title('Quantity vs UnitPrice', fontsize=14)
plt.xlabel('Quantity')
plt.ylabel('Unit Price')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```
