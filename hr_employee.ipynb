```python


# Ensure JobSatisfaction is numeric
df['JobSatisfaction'] = pd.to_numeric(df['JobSatisfaction'], errors='coerce')

# Split data into two groups
yes_attrition = df[df['Attrition'] == 'Yes']['JobSatisfaction']
no_attrition = df[df['Attrition'] == 'No']['JobSatisfaction']

# Perform t-test
t_stat, p_val = stats.ttest_ind(yes_attrition, no_attrition, equal_var=False)

print(f"T-statistic: {t_stat:.2f}, P-value: {p_val:.4f}")
```

```python
df.groupby('Attrition')['JobSatisfaction'].var()
```

```python
#we can reject the null hyptheis becuse there is diffrence in both group
```

```python
df['MonthlyIncome'] = pd.to_numeric(df['MonthlyIncome'], errors='coerce')
yes_attrition = df[df['Attrition'] == 'Yes']['MonthlyIncome']
no_attrition = df[df['Attrition'] == 'No']['MonthlyIncome']

# Perform t-test
t_stat, p_val = stats.ttest_ind(yes_attrition, no_attrition, equal_var=False)

print(f"T-statistic: {t_stat:.2f}, P-value: {p_val:.4f}")
```

```python
#here we can alse reject the null hypothesis
```
