```python
rfm_score=df_customers.groupby('CustomerID').agg(recency=('InvoiceDate', lambda x :(snap_date-x.max()).days),
                                            frequncey=('InvoiceNo','nunique'),
                                                 monetry=('total_1','sum'))
```

```python
rfm_score
```

```python
# Scoring 1-5 like your NTILE — pandas equivalent is qcut (quantile cut)
rfm_score['R_score'] = pd.qcut(rfm_score['recency'], 5, labels=[5,4,3,2,1])  # low recency = best = 5
rfm_score['F_score'] = pd.qcut(rfm_score['frequncey'].rank(method='first'), 5, labels=[1,2,3,4,5])
rfm_score['M_score'] = pd.qcut(rfm_score['monetry'], 5, labels=[1,2,3,4,5])
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
rfm_score.head(2)
```

```python
x=rfm_score[['recency','frequncey','monetry']]
```

```python
s = StandardScaler()
```

```python
s.fit(x)
```

```python
x_scale=s.transform(x)
```

```python
from sklearn.cluster import KMeans
```

```python
elbow=KMeans(n_clusters=4, random_state=42, n_init=10)
```

```python
elbow
```

```python
elbow.fit(x_scale)
```

```python
import matplotlib.pyplot as plt

# Create empty list to store inertia values
inertia_values = []
cluster_range = range(1, 11)  # Try k values from 1 to 10

# Loop through each k value
for k in cluster_range:
    elbow = KMeans(n_clusters=k, random_state=42, n_init=10)
    elbow.fit(x_scale)  # Fit KMeans using your scaled data
    inertia_values.append(elbow.inertia_)  # Save inertia for each k

# Plot the elbow curve
plt.figure(figsize=(8, 5))
plt.plot(cluster_range, inertia_values, marker='o', color='orange')
plt.title('Elbow Method - Optimal Number of Clusters')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Inertia')
plt.show()
```

```python
# Fit the final KMeans model using the chosen number of clusters (4)
elbow = KMeans(n_clusters=4, random_state=42, n_init=10)
elbow.fit(x_scale)

# Assign cluster labels to each customer
rfm_score['Cluster'] = elbow.predict(x_scale)

# View the first few rows to confirm clusters are added
rfm_score.head(1
              )
```

```python
customer_segment=rfm_score.groupby('Cluster').agg(r_mean=('recency','mean'),f_mean=('frequncey','mean'),m_mean=('monetry','mean'))
```

```python
customer_segment.reset_index()
```

```python
customer_segment = customer_segment.reset_index()

customer_segment['labels']=customer_segment['labels'] = customer_segment['Cluster'].map({0: 'occasional', 1: 'At risk', 2: 'Champions', 3: 'loyal'})
```
